```python
import os
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from PIL import Image
from transformers import SegformerForSemanticSegmentation
import matplotlib.pyplot as plt
```

```python
CONFIG = {
    "model_name": "nvidia/segformer-b0-finetuned-cityscapes-512-1024",
    "image_size": (512, 1024),   # H x W
    "num_classes": 19,
    "ignore_index": 255,
    "lr": 6e-5,
    "batch_size": 4,
    "num_epochs": 50,
    "data_root": "./data/cityscapes",   
    "checkpoint_dir": "./checkpoints",
}

os.makedirs(CONFIG["checkpoint_dir"], exist_ok=True)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")
```

```python
ID_MAP = {
    7:0, 8:1, 11:2, 12:3, 13:4,
    17:5, 19:6, 20:7, 21:8, 22:9,
    23:10, 24:11, 25:12, 26:13, 27:14,
    28:15, 31:16, 32:17, 33:18
}

class CityscapesDataset(Dataset):
    def __init__(self, root, split="train"):
        self.image_dir = os.path.join(root, "leftImg8bit", split)
        self.label_dir = os.path.join(root, "gtFine", split)
        self.images = []

        self.image_transform = transforms.Compose([
            transforms.Resize(CONFIG["image_size"]),
            transforms.ToTensor(),
            transforms.Normalize(
                mean=[0.485, 0.456, 0.406],
                std=[0.229, 0.224, 0.225]
            )
        ])

        for city in sorted(os.listdir(self.image_dir)):
            city_img_path = os.path.join(self.image_dir, city)
            for img in sorted(os.listdir(city_img_path)):
                if img.endswith("_leftImg8bit.png"):
                    self.images.append((city, img))

        print(f"[{split}] Found {len(self.images)} images")

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        city, img_name = self.images[idx]

        # Load image
        img_path = os.path.join(self.image_dir, city, img_name)
        image = Image.open(img_path).convert("RGB")
        image = self.image_transform(image)

        # Load mask
        label_name = img_name.replace("_leftImg8bit.png", "_gtFine_labelIds.png")
        label_path = os.path.join(self.label_dir, city, label_name)
        mask = np.array(Image.open(label_path))

        # Map to 19 classes, unmapped → 255
        new_mask = np.ones_like(mask, dtype=np.uint8) * 255
        for k, v in ID_MAP.items():
            new_mask[mask == k] = v

        # Resize mask
        mask_pil = Image.fromarray(new_mask)
        mask_pil = mask_pil.resize(
            (CONFIG["image_size"][1], CONFIG["image_size"][0]),
            Image.NEAREST   # NEAREST — never interpolate labels
        )
        mask_tensor = torch.tensor(np.array(mask_pil), dtype=torch.long)

        return image, mask_tensor
```

```python
train_dataset = CityscapesDataset(CONFIG["data_root"], split="train")
val_dataset   = CityscapesDataset(CONFIG["data_root"], split="val")

train_loader = DataLoader(
    train_dataset,
    batch_size=CONFIG["batch_size"],
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=CONFIG["batch_size"],
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print(f"Train batches: {len(train_loader)}")
print(f"Val batches:   {len(val_loader)}")
```

```python
model = SegformerForSemanticSegmentation.from_pretrained(
    CONFIG["model_name"]
)
model = model.to(device)

optimizer = torch.optim.AdamW(model.parameters(), lr=CONFIG["lr"], weight_decay=0.01)

criterion = nn.CrossEntropyLoss(ignore_index=CONFIG["ignore_index"])

print("Model loaded ✅")
print(f"Parameters: {sum(p.numel() for p in model.parameters()):,}")
```
